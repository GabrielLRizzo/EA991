### Passo 1: carregar os pacotes

```python
""" Importando pacotes usuais """

import numpy as np
import pandas as pd
import matplotlib as mp
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import seaborn as sns
```

### Passo 2: carregar o *dataset*

```python
#Carregando o dataset Iris na forma de um Bunch
iris = load_iris(as_frame=True)

#DataFrame único (inclui atributos e rótulos)
df = iris.frame

labels = {0: 'setosa', 1: 'versicolor',
            2: 'virginica'}

# Acrescenta a coluna 'labels' mapeando os valores de 'target' conforme o dicionário criado
df['labels'] = df['target'].map(labels)

#isola os nomes das classes
class_names = df['labels'].unique()

df.head()
```

| ----- | ----- |
| 0 | Setosa |
